### <font color="blue">Problem 1</font>
In the early days of computing, languages did not provide sort functions.  The developer was required to write the code necessary to sort a collection.  There are many sorting algorithms available. For this problem you are to develop a function called _bubbleSort_ that accepts a list as a parameter and sorts that list using the _bubble sort algorithm_.  A bubble sort is the simplest, and least efficient, sorting algorithm that works by repeatedly swapping the adjacent elements if they are in wrong order.  In effect, the algorithm bubbles larger values to the end of the collection.  The algorithm makes multiple passes over the collection as shown below.

Example: Given the list [5,1,4,2,8]  &nbsp;&nbsp;&nbsp;&nbsp; # thanks to GeeksforGeeks for the example. https://www.geeksforgeeks.org/bubble-sort/

First Pass:  
[**5**,1,4,2,8] –> [1,**5**,4,2,8] &nbsp;&nbsp;&nbsp;&nbsp;# 5 > 1 so swap.  
[1,**5**,4,2,8] –> [1,4,**5**,2,8] &nbsp;&nbsp;&nbsp;&nbsp;# 5 > 4 so swap.  
[1,4,**5**,2,8] –> [1,4,2,**5**,8] &nbsp;&nbsp;&nbsp;&nbsp;# 5 > 2 so swap.   
[1,4,2,**5**,8] –> [1,4,2,**5**,8] &nbsp;&nbsp;&nbsp;&nbsp;# 5 < 8 so do nothing.

Second Pass:   
[1,**4**,2,5,8] –> [1,**4**,2,5,8] &nbsp;&nbsp;&nbsp;&nbsp;# 4 > 1 so do nothing.  
[1,**4**,2,5,8] –> [1,2,**4**,5,8] &nbsp;&nbsp;&nbsp;&nbsp;# 4 > 2 so swap.  
[1,2,**4**,5,8] –> [1,2,**4**,5,8] &nbsp;&nbsp;&nbsp;&nbsp;# 4 < 5 so do nothing.  
[1,2,**4**,5,8] –> [1,2,**4**,5,8] &nbsp;&nbsp;&nbsp;&nbsp;# 4 < 8 so do nothing.   
  
The list is sorted after the second pass but the algorithm doesn't know it.  It has to make a third pass.

Third Pass:   
[1,**2**,4,5,8] –> [1,**2**,4,5,8] &nbsp;&nbsp;&nbsp;&nbsp;# 2 > 1 so do nothing.   
[1,**2**,4,5,8] –> [1,**2**,4,5,8] &nbsp;&nbsp;&nbsp;&nbsp;# 2 < 4 so do nothing.   
[1,**2**,4,5,8] –> [1,**2**,4,5,8] &nbsp;&nbsp;&nbsp;&nbsp;# 2 < 5 so do nothing.     
[1,**2**,4,5,8] –> [1,**2**,4,5,8] &nbsp;&nbsp;&nbsp;&nbsp;# 2 < 8 so do nothing.     

Because the algorithm requires referencing and comparing individual elements, an index is used to reference the elements of the list. The algorithm also requires a nested loop. 

Do the following in the function:
1. Using i as the loop variable, iterate over the range 0 through the length of the list:
   1. Using j as the loop variable, iterate over the range 0 through (length of the list - i - 1):
   2. if the list element referenced by j is greater than the list element reference by j + 1:
      1. Swap the elements
      
Note, your code should work with the list parameter passed to the function, i.e., do not attempt to create a new list.  As evidenced by the above algorithm, the positions of the list elements are changing as the algorithm executes.  The expression _length of list - i - 1_ reduces the upper range since with each successive pass the larger values are being moved to the end of the list.  You may want to use print statements to print i and j to observe how the algorithm works.

```python
def bubbleSort(lst):
    # I use an outer loop to make multiple passes over the list
    for i in range(len(lst)):
        # I use an inner loop to compare adjacent elements, reducing the range each pass
        for j in range(0, len(lst) - i - 1):
            # I swap the two elements if the left one is greater than the right one
            if lst[j] > lst[j + 1]:
                lst[j], lst[j + 1] = lst[j + 1], lst[j]

# Do not change this code
animals = ['elephant', 'cat', 'moose', 'antelope', 'elk', 'rabbit', 'zebra', 'yak', 'salamander', 'deer',\
           'otter', 'minx', 'giraffe', 'goat', 'cow', 'tiger', 'bear']
bubbleSort(animals)
print(animals)
```

Define a function called _sumWord_ that is passed a string argument.  The function returns the sum of the ASCII numeric value of each character in the word.  Use the _ord_ function to determine the ASCII numeric value of a character. For example, if the function is passed the string 'elephant' the function returns 849 which is the sum of each character in the string: e 101, l 108,e 101, p 112, h 104, a 97, n 110, t 116

Use this function and the _sort_ method to sort the list _animals_ below on the ASCII numeric total of each string. 

```python
animals = ['elephant', 'cat', 'moose', 'antelope', 'elk', 'rabbit', 'zebra', 'yak', 'salamander', 'deer',\
           'otter', 'minx', 'giraffe', 'goat', 'cow', 'tiger', 'bear']

def sumWord(word):
    # I use the ord function to get the ASCII value of each character and sum them all up
    return sum(ord(ch) for ch in word)

print("Before the sort:")
print(animals)
print()
# I sort the animals list in place using my sumWord function as the key
animals.sort(key=sumWord)
print("After the sort:")
print(animals)
```

### <font color="blue">Problem 8</font>
Write the function _sortBy_ that accepts two parameters.  The first parameter is the sort key.  The second is a list of movies composed of tuples (year released,movie name,rating).  The function returns a list in similar format but sorted based upon the value of the key parameter. 

The sort key passed to the function determines how the movie list is sorted and may have the following values:

    'year'  which causes to function to sort on the year released.
    'title' which causes to function to sort on the movie title.
    'rating' which causes to function to perform a descending sort on movie rating.    

The function returns _False_ for any value of the sort key other than those listed above.

```python
movies = [
    (1994, 'The Shawshank Redemption', 9.2),
    (1999, 'Fight Club', 8.8),
    (1994, 'Pulp Fiction', 8.9),
    (1972, 'The Godfather', 9.2),
    (2008, 'The Dark Knight', 9.0),
    (2022, 'Avatar: The Way of Water', 7.6),
    (2022, 'A Man Called Otto', 7.5),
    (1972, 'Jeremiah Johnson', 7.6)
]

def sortBy(key, movies):
    # I check the key and return the list sorted by the matching field
    if key == 'year':
        # I sort by index 0 which is the release year
        return sorted(movies, key=lambda x: x[0])
    elif key == 'title':
        # I sort by index 1 which is the movie title
        return sorted(movies, key=lambda x: x[1])
    elif key == 'rating':
        # I sort by index 2 in descending order so the highest rating comes first
        return sorted(movies, key=lambda x: x[2], reverse=True)
    else:
        # I return False if the key is not one of the valid options
        return False

# do not change the code below.
print('Sorted on year:')
print(sortBy('year',movies))
print()
print('Sorted on title:')
print(sortBy('title',movies))
print()
print('Sorted on rating:')
print(sortBy('rating',movies))
print()
print('Sorted on invalid key:')
print(sortBy('xxxx',movies))
print()
```

Write the function _bestScrabbleWords_ that accepts two parameters:  _words_, which is a list of words, and _howMany_, which is an integer with a default value of three.  The function needs to sort the words on their scrabble score and return as list of the top _howMany_ words.  Each element in the return list is a tuple containing the word and its corresponding Scrabble score.  You need to use a lambda function key parameter for the sorted function.

```python
def computeScrabbleScore(word):
    # I define a dictionary mapping each letter to its Scrabble point value
    scores = {"a":1,"e":1,"i":1,"o":1,"u":1,"l":1,"n":1,"r":1,"s":1,"t":1,
              "d":2,"g":2,"b":3,"c":3,"m":3,"p":3,"f":4,"h":4,"v":4,"w":4,
              "y":4,"k":5,"j":8,"x":8,"q":10,"z":10}
    # I sum the score of each character in the word
    return sum(scores.get(ch, 0) for ch in word.lower())

def bestScrabbleWords(words, howMany=3):
    # I build a list of tuples pairing each word with its Scrabble score
    scored = [(word, computeScrabbleScore(word)) for word in words]
    # I sort by score in descending order and return only the top howMany results
    return sorted(scored, key=lambda x: x[1], reverse=True)[:howMany]

# Do not change this code
letter_values =  {'a':1, 'b':3, 'c':3, 'd':2, 'e':1, 'f':4, 'g': 2, 'h':4,
    'i':1, 'j':8, 'k':5, 'l':1, 'm':3, 'n':1, 'o':1, 'p':3, 'q':10, 'r':1,
    's':1, 't':1, 'u':1, 'v':8, 'w':4, 'x':8, 'y':4, 'z':10}
scrabbleWords = ['python', 'program', 'list', 'string', 'unix', 'hours', 'syntax', 'error']
bestWords = bestScrabbleWords(scrabbleWords, 4)
print(bestWords)
```
