### <font color="blue">Problem 1<font>
Write a function named _buildingLatLong_ that is passed a building id, searches _umich_buildings_ for that id, and, if found, and returns a tuple whose elements are (name,latitude, longitude) of the corresponding building. If the building id is not found in _umich_buildings_, the function must return None. 

Write code that iterates over the list _bldgs_, containing building ids, calls _buildingLatLong_, and prints the information as shown. 

```python
bldgs = ['ccrb','aep','taubl','unknown','nq','stadm']

def buildingLatLong(building_id):
    # I search through the umich_buildings list to find the matching building id
    for building in umich_buildings:
        if building['id'] == building_id:
            # I return a tuple of name, latitude, and longitude when I find a match
            return (building['name'], building['lat'], building['lng'])
    # I return None if the building id was not found in the list
    return None

# I iterate over the bldgs list and print the result of each lookup
for b in bldgs:
    print(buildingLatLong(b))
```

### <font color="blue">Problem 2</font>
"Vincenty's formulas are two related iterative methods used to calculate the distance between two points on the surface of a spheroid, developed by Thaddeus Vincenty (1975a)"

The file _Unit08_vincenty.py_, which was imported in the first cell, contains the function _latLngDistanceMeter_ that calculates the _vincenty_ distance" in meters between two points on the Earth using the latitude and longitude of each points. The function accepts two tuples each containing the latitude and longitude of a point and calculates the _vincenty_ distance between those points. Below is an example of how this function is called:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;latLngDistanceMeters((42.3584, -71.0912), (42.278, -83.7382))

Write a function named _buildingDistance_ that is passed two building IDs and determines the vincenty distance between both buildings. The function uses the function _buildingLatLong_, developed in the previous problem, to find the longitude and latidude of each building and calls _latLngDistanceMeters_ to determine and return the _vincenty_ distance.  If one or both building ids are invalid, i.e., buildingLatLong returns None, return _False_ as the distance.

In the main part of your code, iterate over the list _bldgs_ below and call _buildingDistance_ for each tuple in the list. Print the output below.

```python
bldgs = [('ccrb','aep'),('taubl','nq'),('bid1','bid2'),('ccrb','bid1'),('bid2','bid2nq')]

def buildingDistance(id1, id2):
    # I use my buildingLatLong function to get the coordinates of each building
    b1 = buildingLatLong(id1)
    b2 = buildingLatLong(id2)
    # I return False if either building id is invalid (not found)
    if b1 is None or b2 is None:
        return False
    # I pass the lat/lng of both buildings to latLngDistanceMeters and return the result
    return latLngDistanceMeters((b1[1], b1[2]), (b2[1], b2[2]))

# I iterate over the list of building pairs and print the vincenty distance for each
for pair in bldgs:
    dist = buildingDistance(pair[0], pair[1])
    print(f"The vincenty distance between building {pair[0]} and {pair[1]} is {dist}")
```

### <font color="blue">Problem 4</font>
This problem uses the list _baby_names_ loaded from Unit08_us-common-baby-names.csv (see first cell). Each element in the baby_names list is a dictionary.  The keys in the dictionary are 'year', 'name','percent',and 'sex'.  A sample row appears below.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;{'year': '1880', 'name': 'John', 'percent': '0.081541', 'sex': 'boy'}
             
The _percent_ key holds a percentage value, represented as a fraction, which represents the number of babies born that year who were given the name found in the value of the _name_ key.   

Write the function _popularInYear_ that accepts the  parameter _year_ and returns the most popular name for that year and the associated percent. The most popular name is the one with the largest percentage.  This problem is essentially finding the maximum percentage in the _baby_names_ list that matches the _year_ parameter.  

Write code the iterates over the _years_ list and prints the most popular name for each year as shown below.  Note, that you need to multiply the _percent_ value by 100 to print that value as a percentage.

```python
years = ['1900','1925','1950','1975','2000','2004','2008']

def popularInYear(year):
    # I initialize best to None and update it as I find higher percentages
    best = None
    for entry in baby_names:
        # I only consider entries that match the year I am looking for
        if entry['year'] == year:
            if best is None or float(entry['percent']) > float(best['percent']):
                best = entry  # I update best when I find a higher percentage
    return best

# I iterate over each year and print the most popular name and its percentage
for year in years:
    result = popularInYear(year)
    print(f"Most popular name in {year} is {result['name']} with {float(result['percent'])*100:.2f} percent")
```

### <font color="blue">Problem 9</font>
Write a function called _getZipData_ that accepts a _post code_ (zip code) as an argument and returns a tuple containing the associated _place name_, _longitude_, and _latitude_ for _zipCodeData_.  If the _post code_ passed to the function is not in the zipCodeData list, the returned tuple should have a value of None, 0, and 0 for the _place name_, _latitude_, and _longitude_ respectively.

Do the following in the function:
1. Initialize variables as necessary.
2. Iterate over _zipCodeData_: 
   1. Iterate over nexted _places_ list:
      1. If _post code_ in the current place is equal to the _post code_ passed:
         1. Store the _place name, _longitude_, and _latitude in the variables
3. Return the variables as a tuple

```python
def getZipData(post_code):
    # I initialize my return variables to default values in case the zip is not found
    place_name = None
    longitude = 0
    latitude = 0
    # I iterate over each entry in zipCodeData and then over its nested places list
    for entry in zipCodeData:
        for place in entry['places']:
            # I check if the current place's post code matches what I was passed
            if place['post code'] == post_code:
                place_name = place['place name']
                longitude = place['longitude']
                latitude = place['latitude']
    # I return a tuple of place name, longitude, and latitude
    return (place_name, longitude, latitude)

# Do not change the code below
for zipcode in ['49682','49707','11111']:
     print(getZipData(zipcode))
```
