### <font color="blue">Problem 1</font>
The list _inp_ is a list of integers.  Write a while loop that does the following:
1. Iterates over the list _inp_.
2. If the number is greater than 255, terminate the loop.
3. If the number is less than 33 or greater than 126, print the skip message shown.
4. Use the chr function to convert the number to an ASCII printable character.
5. Catenate(join) the character to the string variable _message_.

Print the _message_.

You will need to use an IF-ELIF-ELSE statement.

````python
# iterate over the list inp,
# if number is greater than 255, terminate loop,
# if the number is less than 33 or greater than 126, print skip message
# convert number to character, catenate to message

inp = [71,111,130,32,66,108,117,127,101,33,256,71,111,1032,66,108,117,127,101]
message = ""  # initialize empty string to build the message
i = 0  # index counter for the while loop

# iterate over the list inp,
while i < len(inp):
    num = inp[i]  # get current number from list

    if num > 255:
        break  # stop the loop if number exceeds 255
    elif num < 33 or num > 126:
        print(f"Skipped {num}")  # print skip message for non-printable ASCII
    else:
        message += chr(num)  # convert number to character and add to message

    i += 1  # move to next element

print("The message is", message)
````

### <font color="blue">Problem 4</font>
Complete the function called gcf that calculates the Greatest Common Factor(Divisor) given two integers.  The function accepts two integers x and y as parameters and returns the GCF using the Euclidean algorithm below.  You need to use a while loop.

Step 1: r = x % y<br />
Step 2: x = y<br />
Step 3: y = r<br />
Step 4: Repeat Steps 1 through 3 while y is greater than 0<br />
Step 5: GCF = x<br />

Write code that iterates over the list number_pairs (you may use a for statement) and calls the function gcf with each pair and prints the output shown below.

````python
def gcf(x, y):
    # use Euclidean algorithm to find greatest common factor
    while y > 0:       # keep looping while remainder is not zero
        r = x % y      # step 1: calculate remainder
        x = y          # step 2: replace x with y
        y = r          # step 3: replace y with remainder
    return x           # step 5: x is the GCF when loop ends

# list of number pairs to test the gcf function
number_pairs = [(12,24),(24,48),(17,37)]

# iterate over each pair and print the GCF
for pair in number_pairs:
    result = gcf(pair[0], pair[1])  # call gcf with each pair
    print(f"The greatest common factor of {pair} is {result}")
````

````python
dividend = [32,24,8,16]
divisor = [2,4,0,8]

for d in dividend:          # outer loop: iterate over each dividend
    counter = 0             # reset counter for inner loop each time

    while counter < len(divisor):  # inner loop: iterate over each divisor
        try:
            result = d / divisor[counter]  # attempt division
            print(f"Dividing {d} by {divisor[counter]} has a result of {result}")
        except ZeroDivisionError:
            print(f"Dividing {d} by {divisor[counter]} is undefined")  # handle divide by zero

        counter += 1  # increment counter (outside try/except so it always runs)
````

Write code that creates the comma separated values text file, called _Unit06_wnvhumancases.txt_, by iterating over the dictionary(_wnvDictByYear_) created in the previous problem. Do not be concerned about the order of the records in the file.

You need to use a _while_ loop to iterate over the dictionary.  The example below provides the necessary syntax:

         while wnvDictByYear:
           year, yearDict = wnvDictByYear.popitem()

The _popitem_ method returns the next key/value pair then removes that key/value pair from the dictionary.  This means that once the dictionary is emptied the loop terminates.  The implication is that if your code is rerun it would not produce any output because _wnvDictByYear_ is empty.  You must make your code repeatable without requiring rerunning the previous problem to rebuild _wnvDictByYear_. You may not copy the code from #9 to build _wnvDictByYear_. This is not as evident as you may think. You will need to do some research to find a solution.

````python
import copy

# deep copy the dictionary so popitem() doesn't destroy the original
wnvDictByYearCopy = copy.deepcopy(wnvDictByYear)

f = open('Data/Unit06_wnvhumancases.txt', 'w')  # open output file for writing

while wnvDictByYearCopy:  # loop until dictionary is empty
    year, yearDict = wnvDictByYearCopy.popitem()  # remove and get next year entry

    for county, cases in yearDict.items():  # iterate over each county in that year
        f.write(f"{year},{county},{cases}\n")  # write year, county, cases as CSV line

f.close()  # close the file when done

print("File written successfully.")
````
