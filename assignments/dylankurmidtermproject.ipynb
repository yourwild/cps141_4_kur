## Step 2: Load Inventory Data from CSV

Create a function to read the inventory data from the CSV file and store it in a dictionary. The dictionary should use product names as keys and store quantity and price as values.


```python
def load_inventory(filename):
    inventory = {}

    try:
        # open the file in read mode
        f = open(filename, 'r')
        reader = csv.reader(f)       # create a csv reader object
        next(reader)                 # skip the header row

        for row in reader:
            product_name = row[0]        # first column is product name
            quantity = int(row[1])       # second column is quantity (convert to int)
            price = float(row[2])        # third column is price (convert to float)
            # store in nested dictionary format
            inventory[product_name] = {'quantity': quantity, 'price': price}

        f.close()  # close the file when done

    except FileNotFoundError:
        # if file doesn't exist, print error and return empty dictionary
        print(f"Error: File '{filename}' not found.")

    return inventory

```

## Step 3: Display All Inventory Items

Create a function to display all items in the inventory in a formatted table.


```python
def display_inventory(inventory):
    # print header row with column names
    print(f"{'Product':<30} {'Quantity':<15} {'Price'}")
    print("-" * 55)  # separator line

    # loop through each item and print formatted row
    for product, details in inventory.items():
        print(f"{product:<30} {details['quantity']:<15} ${details['price']:.2f}")
```

## Step 4: Calculate Total Inventory Value

Create a function that calculates the total value of all inventory (quantity × price for each item, then sum all items).


```python
def calculate_total_value(inventory):
    total_value = 0

    # loop through all items and add quantity x price for each
    for product, details in inventory.items():
        total_value += details['quantity'] * details['price']

    return total_value
```

## Step 5: Find Low Stock Items

Create a function that finds all items with quantity below a specified threshold (e.g., 10 items).


```python
def find_low_stock_items(inventory, threshold=10):
    low_stock = []

    # loop through inventory and check each item's quantity
    for product, details in inventory.items():
        if details['quantity'] < threshold:  # below threshold means low stock
            low_stock.append(product)        # add product name to list

    return low_stock
```

## Step 6: Add or Update Inventory Item

Create a function that allows adding a new product or updating an existing product's quantity or price.


```python
def add_or_update_item(inventory, product_name, quantity, price):
    if product_name in inventory:
        # product already exists: update quantity and price
        inventory[product_name]['quantity'] = quantity
        inventory[product_name]['price'] = price
        print(f"'{product_name}' has been updated.")
    else:
        # product is new: add it to the inventory
        inventory[product_name] = {'quantity': quantity, 'price': price}
        print(f"'{product_name}' has been added.")
```

## Step 7: Find Most and Least Expensive Items

Create functions to find the most expensive and least expensive items in the inventory.


```python
def find_most_expensive(inventory):
    if not inventory:  # return default if inventory is empty
        return (None, 0)

    max_product = None
    max_price = -1  # start low so any price beats it

    for product, details in inventory.items():
        if details['price'] > max_price:  # found a higher price
            max_price = details['price']
            max_product = product

    return (max_product, max_price)


def find_least_expensive(inventory):
    if not inventory:  # return default if inventory is empty
        return (None, 0)

    min_product = None
    min_price = float('inf')  # start very high so any price beats it

    for product, details in inventory.items():
        if details['price'] < min_price:  # found a lower price
            min_price = details['price']
            min_product = product

    return (min_product, min_price)

```

## Step 8: Generate Inventory Report

Create a function that generates a comprehensive inventory report with statistics.


```python
def generate_report(inventory):
    print("=" * 55)
    print("INVENTORY REPORT")
    print("=" * 55)

    # total number of products
    print(f"Total number of products:  {len(inventory)}")

    # total inventory value
    total_value = calculate_total_value(inventory)
    print(f"Total inventory value:     ${total_value:.2f}")

    # most expensive item
    most_exp_name, most_exp_price = find_most_expensive(inventory)
    print(f"Most expensive item:       {most_exp_name} (${most_exp_price:.2f})")

    # least expensive item
    least_exp_name, least_exp_price = find_least_expensive(inventory)
    print(f"Least expensive item:      {least_exp_name} (${least_exp_price:.2f})")

    # low stock items
    low_stock = find_low_stock_items(inventory)
    print(f"Low stock items:           {low_stock if low_stock else 'None'}")

    # average price
    avg_price = calculate_average_price(inventory)
    print(f"Average price:             ${avg_price:.2f}")

    # total quantity
    total_qty = calculate_total_quantity(inventory)
    print(f"Total quantity (all items): {total_qty}")

    print("=" * 55)
```

## Step 9: Search for Product

Create a function to search for a product in the inventory by name.


```python
def search_product(inventory, product_name):
    # case-insensitive search by comparing lowercase versions
    for product, details in inventory.items():
        if product.lower() == product_name.lower():
            # return matching product in same dictionary format
            return {product: {'quantity': details['quantity'], 'price': details['price']}}

    return None  # not found
```

## Step 10: Remove Item from Inventory

Create a function to remove items from the inventory.


```python
def remove_item(inventory, product_name):
    if product_name in inventory:
        del inventory[product_name]  # remove the product
        print(f"'{product_name}' has been removed from inventory.")
    else:
        print(f"'{product_name}' was not found in inventory.")
```

## Step 11: Price Range Filter

Create a function to find all items within a specific price range.


```python
def find_items_in_price_range(inventory, min_price, max_price):
    results = []  # list to store matching items

    for product, details in inventory.items():
        price = details['price']
        # check if price falls within the specified range (inclusive)
        if min_price <= price <= max_price:
            results.append((product, price))  # add tuple to results

    return results
```

## Step 13: Interactive Menu System

This interactive menu system using a while loop that allows users to:
- View inventory
- Add items
- Update items
- Remove items
- Search for products
- View reports
- Exit and Clear Past Entries


```python
from IPython.display import display, clear_output
import ipywidgets as widgets

def run_interactive_menu(inventory):
    """
    Run an interactive menu system using IPython widgets.
    This avoids the clear_output/input prompt issues in Jupyter.
    """
    output = widgets.Output()

    def handle_menu_choice(choice):
        with output:
            clear_output(wait=True)


            if choice == '1':
                print("\n" + "="*60)
                print("COMPLETE INVENTORY")
                print("="*60)
                display_inventory(inventory)
                print("="*60)

            elif choice == '2':
                print("\n" + "="*60)
                print("ADD/UPDATE ITEM")
                print("="*60)

                # Create input widgets
                name_input = widgets.Text(description='Product Name:', style={'description_width': 'initial'}, layout=widgets.Layout(width='400px'))
                qty_input = widgets.IntText(description='Quantity:', style={'description_width': 'initial'}, layout=widgets.Layout(width='400px'))
                price_input = widgets.FloatText(description='Price:', style={'description_width': 'initial'}, layout=widgets.Layout(width='400px'))
                submit_btn = widgets.Button(description='Submit', button_style='success')
                cancel_btn = widgets.Button(description='Cancel', button_style='')

                def on_submit(b):
                    with output:
                        clear_output(wait=True)
                        print("\n" + "="*60)
                        print("ITEM ADDED/UPDATED SUCCESSFULLY")
                        print("="*60)
                        add_or_update_item(inventory, name_input.value, qty_input.value, price_input.value)
                        print("="*60)

                def on_cancel(b):
                    with output:
                        clear_output(wait=True)


                submit_btn.on_click(on_submit)
                cancel_btn.on_click(on_cancel)

                display(widgets.VBox([name_input, qty_input, price_input,
                                      widgets.HBox([submit_btn, cancel_btn])]))
                return

            elif choice == '3':
                print("\n" + "="*60)
                print("REMOVE ITEM")
                print("="*60)

                name_input = widgets.Text(description='Product Name:', style={'description_width': 'initial'}, layout=widgets.Layout(width='400px'))
                submit_btn = widgets.Button(description='Remove', button_style='danger')
                cancel_btn = widgets.Button(description='Cancel', button_style='')

                def on_submit(b):
                    with output:
                        clear_output(wait=True)
                        print("\n" + "="*60)
                        remove_item(inventory, name_input.value)
                        print("="*60)

                def on_cancel(b):
                    with output:
                        clear_output(wait=True)


                submit_btn.on_click(on_submit)
                cancel_btn.on_click(on_cancel)

                display(widgets.VBox([name_input, widgets.HBox([submit_btn, cancel_btn])]))
                return

            elif choice == '4':
                print("\n" + "="*60)
                print("SEARCH FOR PRODUCT")
                print("="*60)

                name_input = widgets.Text(description='Product Name:', style={'description_width': 'initial'}, layout=widgets.Layout(width='400px'))
                search_btn = widgets.Button(description='Search', button_style='info')
                cancel_btn = widgets.Button(description='Cancel', button_style='')

                def on_search(b):
                    with output:
                        clear_output(wait=True)

                        result = search_product(inventory, name_input.value)
                        if result:
                            print("\n" + "=" * 60)
                            print("PRODUCT FOUND")
                            print("=" * 60)
                            display_inventory(result)
                            print("=" * 60)
                        else:
                            print("\n" + "=" * 60)
                            print(f"PRODUCT '{name_input.value}' NOT FOUND.")
                            print("=" * 60)

                def on_cancel(b):
                    with output:
                        clear_output(wait=True)


                search_btn.on_click(on_search)
                cancel_btn.on_click(on_cancel)

                display(widgets.VBox([name_input, widgets.HBox([search_btn, cancel_btn])]))
                return

            elif choice == '5':
                print("\n" + "=" * 60)
                print("INVENTORY REPORT")
                print("=" * 60)
                generate_report(inventory)
                print("=" * 60)

            elif choice == '6':
                print("\n" + "="*60)
                print("FIND ITEMS IN PRICE RANGE")
                print("="*60)

                min_input = widgets.FloatText(description='Min Price:', style={'description_width': 'initial'}, layout=widgets.Layout(width='400px'))
                max_input = widgets.FloatText(description='Max Price:', style={'description_width': 'initial'}, layout=widgets.Layout(width='400px'))
                search_btn = widgets.Button(description='Search', button_style='info')
                cancel_btn = widgets.Button(description='Cancel', button_style='')

                def on_search(b):
                    with output:
                        clear_output(wait=True)

                        results = find_items_in_price_range(inventory, min_input.value, max_input.value)
                        if results:
                            print("\n" + "=" * 60)
                            print("FOUND ITEMS")
                            print("=" * 60)
                            print(f"\nItems in price range ${min_input.value:.2f} - ${max_input.value:.2f}:")
                            for product_name, price in results:
                                print(f"  {product_name}: ${price:.2f}")
                            print("="*60)
                        else:
                            print("\n" + "=" * 60)
                            print("NO ITEMS FOUND IN THAT PRICE RANGE")
                            print("=" * 60)

                def on_cancel(b):
                    with output:
                        clear_output(wait=True)


                search_btn.on_click(on_search)
                cancel_btn.on_click(on_cancel)

                display(widgets.VBox([min_input, max_input, widgets.HBox([search_btn, cancel_btn])]))
                return



    # Create menu dropdown
    menu_dropdown = widgets.Dropdown(
        options=[
            ('Select an option...', ''),
            ('1. View Inventory', '1'),
            ('2. Add/Update Item', '2'),
            ('3. Remove Item', '3'),
            ('4. Search for Product', '4'),
            ('5. View Report', '5'),
            ('6. Find Items in Price Range', '6')

        ],
        description='Menu:',
        value='',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='400px')
    )

    # Handle menu selection
    def on_menu_change(change):
        if change['new']:  # Only process if a selection was made
            handle_menu_choice(change['new'])
            menu_dropdown.value = ''  # Reset dropdown

    menu_dropdown.observe(on_menu_change, names='value')

    # Display the menu and output area
    display(widgets.VBox([menu_dropdown, output]))
```

**Question 1:** Why is it important to use try-except blocks when reading from and writing to CSV files? What specific exceptions should you handle, and what could go wrong if you don't handle them?

**Answer:** Using try-except blocks is really important when working with CSV files because a lot of things can go wrong that you cant always predict. The biggest one is FileNotFoundError, which happens when the file path is wrong or the file just isnt there yet. If you dont catch that, the whole program just crashes with a weird error message that doesnt help the user at all. Other things that can go wrong include not having permission to open the file, or the data inside being formatted wrong so it cant be converted properly. By wrapping the file reading code in a try-except, the program can handle those situations gracefully and let the user know what happened instead of just stopping.

**Question 2:** What are the advantages of using Python's csv module to read CSV files instead of reading the file as plain text and splitting by commas? Provide a specific example where the csv module would handle data correctly but plain text reading would fail.

**Answer:** The csv module is way more reliable than just doing line.split(',') on every line. The biggest problem with plain text splitting is that it breaks as soon as a value contains a comma. For example if you had a product called "Hard Drive, External" and you tried to split that line by comma, you would end up with too many pieces and your indexes would be completely off. The csv module knows to look for quoted fields and treats everything inside the quotes as one single value, so it handles that situation perfectly. It also deals with things like newlines inside fields and different types of delimiters, which plain text reading just cant do.

**Question 3:** When reading data from a CSV file, all values are initially strings. Explain why it's necessary to convert the quantity to an int and the price to a float. What would happen if you tried to perform mathematical operations without converting these values first?

**Answer:** When Python reads a CSV file every single value comes in as a string, even if it looks like a number. So something like 15 is actually the string "15" and 899.99 is the string "899.99". If you try to multiply those together to get a total value, Python will throw a TypeError because you cant do math on strings. Even something like adding two quantities together would give you the wrong answer since "15" + "5" in Python gives you "155" instead of 20 because it just concatenates the strings. Converting quantity to int and price to float right when you read the data makes sure all the math in the rest of the program works the way it should.

**Question 4:** Explain the benefits of breaking the inventory management system into separate functions rather than writing everything in a single main() function. How does this approach make the code easier to test, debug, and maintain?

**Answer:** Breaking everything into separate functions makes the project a lot easier to work with. It also makes the code much easier to read because each function has one clear job and you can tell what it does just from its name. Another big advantage is that you can reuse functions in multiple places. For example calculate_total_value gets called inside generate_report but could also be called anywhere else without having to rewrite that logic. If I had put everything in one big main function it would have been really hard to find bugs or make changes without breaking something else.

**Question 5:** Explain why a dictionary is the most appropriate data structure for storing the inventory data, compared to using a list or tuple. What advantages does a dictionary provide for this specific use case?

**Answer:** A dictionary made the most sense for storing the inventory because the whole point is to look things up by product name. With a dictionary you can just write inventory["Laptop"] and instantly get all the information about that product without having to loop through anything. If I had used a list I would have had to search through every single item every time I wanted to find one product, which gets slow as the inventory grows. A tuple wasnt even really an option because tuples are immutable, meaning you cant change the values once they are set, and an inventory system obviously needs to be able to update quantities and prices. The nested dictionary format where each product name maps to another dictionary with quantity and price also keeps the data really organized and easy to understand.
