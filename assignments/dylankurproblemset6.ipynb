### <font color="blue">Problem 1</font>
The list _inp_ is a list of integers.  Write a while loop that does the following:
1. Iterates over the list _inp_.
2. If the number is greater than 255, terminate the loop.
3. If the number is less than 33 or greater than 126, print the skip message shown.
4. Use the chr function to convert the number to an ASCII printable character.
5. Catenate(join) the character to the string variable _message_.

Print the _message_.

You will need to use an IF-ELIF-ELSE statement.

```python
inp = [71,111,130,32,66,108,117,127,101,33,256,71,111,1032,66,108,117,127,101]
message = ""  # initialize empty string to build the message
i = 0  # index counter for the while loop

while i < len(inp):
    num = inp[i]  # get current number from list

    if num > 255:
        break  # stop the loop if number exceeds 255
    elif num < 33 or num > 126:
        print(f"Skipped {num}")  # print skip message for non-printable ASCII
    else:
        message += chr(num)  # convert number to character and add to message

    i += 1  # move to next element

print("The message is", message)
```

Write a code that calculates the largest, smallest and average score in the set of entered scores. 

Do the following:
1. Initialize a variable to an empty list.  This variable will hold the scores.
2. Initialize a counter variable.
3. Accept the number of scores from the user.
4. Iterate continuously:
   1. If the counter is less than the number of scores:
        1. Accept a score from the user.
        2. Append the score to the scores list.
        3. increment the counter.
   2. Otherwise:
        1. Calculate the largest score (you may use a function).
        2. Calculate the smallest score (you may use a function).

6. Print the average, smallest, and largest score as shown below.

```python
scores = []     # list to hold all entered scores
counter = 0     # counts how many scores have been entered
num_scores = int(input("How many scores are you entering? "))  # get total count from user

while True:  # loop continuously
    if counter < num_scores:
        score = float(input("Add another number: "))  # get score from user
        scores.append(score)   # add score to list
        counter += 1           # increment counter
    else:
        largest = max(scores)   # find the largest score
        smallest = min(scores)  # find the smallest score
        break  # all scores entered, exit loop

average = sum(scores) / len(scores)  # calculate the average

print(f"The average score is: {average}")
print(f"The smallest score is: {int(smallest)}")
print(f"The largest score is: {int(largest)}")
```

### <font color="blue">Problem 4</font>
Complete the function called gcf that calculates the Greatest Common Factor(Divisor) given two integers.  The function accepts two integers x and y as parameters and returns the GCF using the Euclidean algorithm below.  You need to use a while loop.

Step 1: r = x % y<br />
Step 2: x = y<br />
Step 3: y = r<br />
Step 4: Repeat Steps 1 through 3 while y is greater than 0<br />
Step 5: GCF = x<br />

Write code that iterates over the list number_pairs (you may use a for statement) and calls the function gcf with each pair and prints the output shown below.

```python
def gcf(x, y):
    # use Euclidean algorithm to find greatest common factor
    while y > 0:       # keep looping while remainder is not zero
        r = x % y      # step 1: calculate remainder
        x = y          # step 2: replace x with y
        y = r          # step 3: replace y with remainder
    return x           # step 5: x is the GCF when loop ends

# Do not change the line below
number_pairs = [(12,24),(24,48),(17,37)]

# iterate over each pair and print the GCF
for pair in number_pairs:
    result = gcf(pair[0], pair[1])  # call gcf with each pair
    print(f"The greatest common factor of {pair} is {result}")
```

The problem requires using the function _computeScrabbleScore_ developed in an earlier problem set.  Copy the code for the function to the cell below.

Write code that repeatedly prompts the user for a word until the sentinel "None" or "none" is entered. In the loop, call _computeScrabbleScore_ with the word entered then add the word and its value as a key-value pair into the dictionary _scrabbleWords_. Print _scrabbleWords_.

```python
def computeScrabbleScore(word):
    # dictionary mapping each letter to its Scrabble point value
    scores = {"a": 1, "e": 1, "i": 1, "o": 1, "u": 1, "l": 1, "n": 1, "r": 1, "s": 1, "t": 1,
              "d": 2, "g": 2,
              "b": 3, "c": 3, "m": 3, "p": 3,
              "f": 4, "h": 4, "v": 4, "w": 4, "y": 4,
              "k": 5,
              "j": 8, "x": 8,
              "q": 10, "z": 10}
    return sum(scores.get(ch, 0) for ch in word.lower())  # sum points for each letter

scrabbleWords = {}  # dictionary to store word:score pairs

while True:  # keep prompting until sentinel is entered
    word = input("Please input a word, enter 'None' or 'none' to cancel: ")

    if word == 'None' or word == 'none':
        break  # exit loop on sentinel value

    scrabbleWords[word.lower()] = computeScrabbleScore(word)  # store word and its score

print(scrabbleWords)
```

Do the following:
1. Iterate over each value in _dividend_:
   1. Initialize a counter
   2. Iterate while the counter is less than the length of _divisor_:
      1. Try:
         1. Divided the current dividend (from the outer loop) by the element in _divisor_ referenced by the counter.
         2. Print the result as shown below.
      2. Except:
         1. Print the divide by 0 result as shown below.
   3. Increment the counter

```python
dividend = [32,24,8,16]
divisor = [2,4,0,8]

for d in dividend:          # outer loop: iterate over each dividend
    counter = 0             # reset counter for inner loop each time

    while counter < len(divisor):  # inner loop: iterate over each divisor
        try:
            result = d / divisor[counter]  # attempt division
            print(f"Dividing {d} by {divisor[counter]} has a result of {result}")
        except ZeroDivisionError:
            print(f"Dividing {d} by {divisor[counter]} is undefined")  # handle divide by zero

        counter += 1  # increment counter (outside try/except so it always runs)
```

You are to write a password generator.  You code needs to do the following:

1. Prompt the user for a password length.
2. Use a while loop to iterate up to the password length.
3. In the loop generate a random character within the ASCII printable range.
4. Accumulate the character in the password.
5. After the loop terminates, print the password.

To generate a random character, you will need to import the _random_ module and use the _random.randrange_ function to generate a number between 48 and 122 (printable ASCII character range).  Convert the number to a character using the _chr_ function.  Accumulate that character in the password variable.

```python
import random

# initialize the pseudorandom number generator
random.seed()

print("Password Generator")
length = int(input("How many characters would you like your password to be? "))

password = ""  # initialize empty password string
counter = 0    # counter to track how many characters have been added

while counter < length:  # loop until password reaches desired length
    num = random.randrange(48, 123)  # generate random number in printable ASCII range
    password += chr(num)             # convert number to character and add to password
    counter += 1                     # increment counter

print(password)
```

### <font color="blue">Problem 10</font>
Write code that creates the comma separated values text file, called _Unit06_wnvhumancases.txt_, by iterating over the dictionary(_wnvDictByYear_) created in the previous problem.  Each line in the file must consist of three columns:  year, county, and number of cases for that county.  Below is an example showing some of the records.  Do not be concerned about the order of the records in the file.

        2018,Amador,1
        2018,Butte,6
        2018,Contra Costa,1
        ...
        2006,Tulare,8
        2006,Ventura,3
        2006,Yolo,26
        2006,Yuba,5


You need to use a _while_ loop to iterate over the dictionary.  The example below provides the necessary syntax:

         while wnvDictByYear:
           year, yearDict = wnvDictByYear.popitem()

The _popitem_ method returns the next key/value pair then removes that key/value pair from the dictionary.  This means that once the dictionary is emptied the loop terminates.  The implication is that if your code is rerun it would not produce any output because _wnvDictByYear_ is empty.  You must make your code repeatable without requiring rerunning the previous problem to rebuild _wnvDictByYear_. You may not copy the code from #9 to build _wnvDictByYear_. This is not as evident as you may think. You will need to do some research to find a solution.

```python
import copy

# deep copy the dictionary so popitem() doesn't destroy the original
wnvDictByYearCopy = copy.deepcopy(wnvDictByYear)

f = open('Data/Unit06_wnvhumancases.txt', 'w')  # open output file for writing

while wnvDictByYearCopy:  # loop until dictionary is empty
    year, yearDict = wnvDictByYearCopy.popitem()  # remove and get next year entry

    for county, cases in yearDict.items():  # iterate over each county in that year
        f.write(f"{year},{county},{cases}\n")  # write year, county, cases as CSV line

f.close()  # close the file when done

print("File written successfully.")
```
